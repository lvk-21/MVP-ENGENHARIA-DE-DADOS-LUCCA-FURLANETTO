# 01 - Ingestão dos Dados da PRF

## Objetivo

Realizar a leitura dos arquivos CSV de acidentes da Polícia Rodoviária Federal referentes aos anos de 2021 a 2025, armazenados no volume `prf_raw` do Databricks.

Nesta etapa, os dados serão lidos em seu formato bruto, sem aplicação de transformações, com o objetivo de preservar a estrutura original fornecida pela fonte.

## Fonte dos arquivos

Portal de Dados Abertos da Polícia Rodoviária Federal (PRF).

## Período analisado

2021 a 2025.

```python
caminho = "/Volumes/workspace/default/prf_raw"

df_2025 = (
    spark.read
    .option("header", True)
    .option("sep", ";")
    .option("encoding", "windows-1252")
    .csv(f"{caminho}/datatran2025.csv")
)
```

```python
df_2025.count()
```

```python
df_2025.printSchema()
```

```python
display(df_2025.limit(20))
```

### Validação da leitura inicial

O arquivo `datatran2025.csv` foi lido com sucesso a partir do volume `prf_raw`.

Foram identificados 72.529 registros.

Nesta etapa, todas as colunas foram mantidas como `string`, evitando inferência automática de tipos. Essa decisão foi tomada para preservar a estrutura original dos dados durante a ingestão bruta e permitir que as conversões de tipo sejam realizadas posteriormente de forma controlada durante o processo de transformação.

```python
anos = [2021, 2022, 2023, 2024, 2025]

dfs = {}

for ano in anos:
    dfs[ano] = (
        spark.read
        .option("header", True)
        .option("sep", ";")
        .option("encoding", "windows-1252")
        .csv(f"{caminho}/datatran{ano}.csv")
    )

for ano in anos:
    print(f"{ano}: {len(dfs[ano].columns)} colunas")
    print(dfs[ano].columns)
    print("-" * 80)
```

```python
for ano in anos:
    print(f"{ano}: {dfs[ano].count()} registros")
```

```python
colunas_referencia = dfs[2021].columns

for ano in anos[1:]:
    if dfs[ano].columns == colunas_referencia:
        print(f"{ano}: mesmo conjunto e ordem de colunas de 2021")
    else:
        print(f"{ano}: estrutura diferente de 2021")
```

## Validação da Estrutura dos Arquivos

Após a leitura dos arquivos referentes aos anos de 2021 a 2025, foi realizada uma validação inicial da estrutura dos dados.

Todos os arquivos apresentaram **30 colunas**, com o mesmo conjunto de campos e na mesma ordem. Isso indica que a estrutura dos arquivos permaneceu consistente ao longo do período analisado.

A quantidade de registros identificada em cada arquivo foi:

- 2021: 64.567 registros
- 2022: 64.606 registros
- 2023: 67.766 registros
- 2024: 73.156 registros
- 2025: 72.529 registros

Também foi realizada uma comparação automática entre as colunas de cada ano, utilizando o arquivo de 2021 como referência. Os arquivos de 2022, 2023, 2024 e 2025 apresentaram o mesmo conjunto e a mesma ordem de colunas.

Essa validação é importante porque garante que os arquivos podem ser combinados em um único conjunto de dados sem necessidade de ajustes estruturais prévios.

Nesta etapa, os dados ainda permanecem em seu formato bruto, sem transformações ou conversões de tipo.

```python
from functools import reduce
from pyspark.sql.functions import lit

dfs_com_metadados = []

for ano in anos:
    df_temp = (
        dfs[ano]
        .withColumn("ano_arquivo", lit(ano))
        .withColumn("arquivo_origem", lit(f"datatran{ano}.csv"))
    )
    
    dfs_com_metadados.append(df_temp)

df_bronze = reduce(
    lambda df1, df2: df1.unionByName(df2),
    dfs_com_metadados
)
```

```python
df_bronze.groupBy(
    "ano_arquivo",
    "arquivo_origem"
).count().orderBy("ano_arquivo").show()
```

```python
display(df_bronze.limit(20))
```

## Consolidação dos Arquivos

Após a validação da estrutura dos cinco arquivos anuais, os dados referentes ao período de 2021 a 2025 foram consolidados em um único DataFrame.

Antes da união, foram adicionadas duas colunas de controle:

- `ano_arquivo`: identifica o ano correspondente ao arquivo de origem;
- `arquivo_origem`: registra o nome do arquivo CSV do qual cada registro foi obtido.

Esses campos foram adicionados para aumentar a rastreabilidade dos dados ao longo do pipeline e facilitar a identificação da origem de cada registro.

A consolidação foi realizada utilizando `unionByName`, uma vez que todos os arquivos apresentaram o mesmo conjunto e a mesma ordem de colunas.

Após a união, o conjunto consolidado apresentou **342.624 registros e 32 colunas**, sendo 30 provenientes originalmente da base da PRF e 2 colunas adicionais de controle.

A distribuição de registros por arquivo de origem foi validada após a consolidação, confirmando que nenhum registro foi perdido durante o processo.

Nesta etapa, nenhuma transformação foi aplicada aos dados originais da PRF. Os dados permanecem em sua forma bruta, acrescidos apenas dos metadados necessários para rastreabilidade.

```python
(
    df_bronze.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.default.prf_acidentes_bronze")
)
```

## Persistência da Camada Bronze

Após a consolidação dos arquivos de 2021 a 2025, o DataFrame resultante foi persistido no Databricks como uma tabela Delta denominada `workspace.default.prf_acidentes_bronze`.

A persistência foi realizada para que os dados brutos consolidados permanecessem disponíveis independentemente da execução do notebook e pudessem ser reutilizados nas etapas seguintes do pipeline.

Foi utilizado o formato Delta Lake, formato de tabela utilizado nativamente pelo Databricks e que adiciona recursos de confiabilidade e gerenciamento sobre os dados armazenados.

Após a gravação, a tabela foi validada por meio de nova leitura e consultas SQL.

A tabela Bronze apresentou:

- 342.624 registros;
- 32 colunas;
- 30 colunas provenientes originalmente dos arquivos da PRF;
- 2 colunas adicionais de rastreabilidade: `ano_arquivo` e `arquivo_origem`.

Os campos originais permaneceram sem transformação nesta etapa, preservando os dados conforme recebidos da fonte.
